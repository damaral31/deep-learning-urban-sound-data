```python
from models.CNN import SoundCNN
from models.BiRNN import SoundBiRNN
from models.RNN import SoundRNN
from models.LSTM import SoundLSTM
from trainer import Train
```

```python
FOLDS = [f"fold{i}" for i in range(1, 11)]
DATA_PATH = "datasets"
```

```python
model_instance = SoundLSTM(num_classes=10, in_channels=1)
trainer = Train(dataset_root_path=DATA_PATH, name="LSTM", dataset_type="singlechannel", model=model_instance, num_classes=10, batch_size=64, epochs=50, learning_rate=1e-3, patience=5)
trainer.run(folds=FOLDS)
```

```python
model_instance = SoundCNN(num_classes=10, SqueezeExcitation=True, AttentionBlock=True, in_channels=5)
trainer = Train(dataset_root_path=DATA_PATH, name="ASECNN", dataset_type="preprocessing", model=model_instance, num_classes=10, batch_size=64, epochs=50, learning_rate=1e-3, patience=5)
trainer.run(folds=FOLDS)
```

```python
model_instance = SoundCNN(num_classes=10, SqueezeExcitation=False, AttentionBlock=False, in_channels=1)
trainer = Train(dataset_root_path=DATA_PATH, name="CNN", dataset_type="singlechannel", model=model_instance, num_classes=10, batch_size=64, epochs=50, learning_rate=1e-3, patience=5)
trainer.run(folds=FOLDS)
```

```python
model_instance = SoundRNN(num_classes=10, in_channels=1)
trainer = Train(dataset_root_path=DATA_PATH, name="RNN", dataset_type="singlechannel", model=model_instance, num_classes=10, batch_size=64, epochs=50, learning_rate=1e-3, patience=5)
trainer.run(folds=FOLDS)
```

```python
model_instance = SoundBiRNN(num_classes=10, in_channels=1, input_height=64)
trainer = Train(dataset_root_path=DATA_PATH, name="BiRNN", dataset_type="singlechannel", model=model_instance, num_classes=10, batch_size=64, epochs=50, learning_rate=1e-3, patience=8)
trainer.run(folds=FOLDS)
```
